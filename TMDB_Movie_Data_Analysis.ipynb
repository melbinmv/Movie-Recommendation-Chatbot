```python
# Libraries
import pandas as pd
```

```python
#get TMDB dataset from drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#Read TMDB dataset from drive
movie_dat = pd.read_csv('/content/drive/MyDrive/dev/TMDB_movie_dataset_v11.csv')
movie_dat
```

```python
# Dataset columns
movie_dat.columns
```

```python
# Selecting the columns which are needed for the analysis
movie_dat = movie_dat[['id','title','vote_average','vote_count','release_date','overview','original_language','genres','keywords','tagline','revenue','budget','status','popularity']]
movie_dat
```

```python
movie_dat.info()
```

```python
# Finding the null values
movie_dat.isnull().sum()
```

```python
# Show the movies whose title is null
movie_dat[movie_dat['title'].isnull()]
# Delete all the titles with null values
movie_dat = movie_dat.dropna(subset=['title'])
```

```python
# Re-checking for null values again in the title
movie_dat.isnull().sum()
```

```python
# Creating a new dataframe to store movies that are not in english
movie_not_english = movie_dat[movie_dat['original_language'] != 'en']
movie_not_english
```

```python
# Remove movies with titles not in english from original dataframe
movie_dat = movie_dat[movie_dat['original_language'] == 'en']
movie_dat
```

```python
# Drop movies with overview is None
movie_not_english = movie_not_english.dropna(subset=['overview'])
#Drop movies with status Planned, Rumored and Canceled
movie_not_english.drop(movie_not_english[movie_not_english['status'].isin(['Planned', 'Rumored', 'Canceled'])].index, inplace=True)
# Drop movies with release date is None
movie_not_english = movie_not_english.dropna(subset=['release_date'])
```

```python
movie_not_english.isnull().sum()
```

```python
movie_not_english.shape
```

```python
# Checking for null values again
movie_dat.isnull().sum()
```

```python
#Delete the movies with overview is None
movie_dat = movie_dat.dropna(subset=['overview'])
movie_dat
```

```python
#Show movies with missing release date
movie_dat[movie_dat['release_date'].isnull()]
```

```python
#Find the unique values in the column status
movie_dat['status'].unique()
```

```python
#Drop movies with status Planned, Rumored and Canceled
movie_dat.drop(movie_dat[movie_dat['status'].isin(['Planned', 'Rumored', 'Canceled'])].index, inplace=True)
movie_dat
```

```python
movie_dat.isnull().sum()
```

```python
movie_with_nodates = movie_dat[movie_dat['release_date'].isna()]
print(f"Movies with missing release dates: {len(movie_with_nodates)} ({len(movie_with_nodates)/len(movie_dat)*100:.2f}%)")
```

```python
movie_dat = movie_dat[~movie_dat['release_date'].isna()]
movie_dat
```

```python
movie_dat.isnull().sum()
```

```python
# Sort These Movies by Popularity (Ascending Order)
movie_with_nodates = movie_with_nodates.sort_values(by="popularity", ascending=True)
movie_with_nodates
```

```python
movie_with_nodates['null_count'] = movie_with_nodates.isna().sum(axis=1)
movies_with_most_nulls= movie_with_nodates.sort_values(by="null_count", ascending=False)
```

```python
movies_with_most_nulls.shape
```

```python
movies_with_most_nulls
```

```python
# Number of movies with null_count is 4
movies_with_most_nulls[movies_with_most_nulls['null_count'] == 4]
```

```python
# Drop the movies with null_count = 4
movie_with_nodates = movie_with_nodates.drop(movies_with_most_nulls[movies_with_most_nulls['null_count'] == 4].index)
movie_with_nodates
```

```python
print(f"Movies with missing release dates: {len(movie_with_nodates)} ({len(movie_with_nodates)/len(movie_dat)*100:.2f}%)")
```

```python
# Merge the Remaining Movies Back into the Main Dataset
movie_dat = pd.concat([movie_dat, movie_with_nodates, movie_not_english], ignore_index=True)
movie_dat
```

```python
movie_dat.isnull().sum()
```

```python
# Show Duplicates
movie_dat[movie_dat.duplicated()]
```

```python
# Drop duplicates
movie_dat = movie_dat.drop_duplicates()
```

```python
#Create a new category "Unknown" for missinge genres
movie_dat['genres'].fillna('Unknown', inplace=True)
```

```python
movie_dat.isnull().sum()
```

```python
#Drop column null count
movie_dat.drop(columns=['null_count'], inplace=True)
```

```python
movie_dat.isnull().sum()
```

```python
movie_dat.shape
```

```python
# Drop column tagline as large portion is missing
movie_dat.drop(columns=['tagline'], inplace=True)
```

```python
movie_dat.isnull().sum()
```

```python
# Export cleaned data
movie_dat.to_csv('/content/drive/MyDrive/dev/TMDB_movie_dataset_v12.csv', index=False)
```
